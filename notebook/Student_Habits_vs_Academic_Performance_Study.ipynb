```python
plt.hist(df["exam_score"], bins=10)
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Scores")
plt.ylabel("Number of Students")
plt.show()
print("Insight:Most students scored in the highest frequency score range shown by the tallest bar. This indicates that the majority of students performed within this score interval.")
```

```python
plt.scatter(df["study_hours_per_day"], df["exam_score"])
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.show()
print("Insight:The scatter plot shows a positive relationship between study hours and exam scores. Students who studied more generally tended to achieve higher marks.")
```

```python
avg_score = df.groupby("gender")["exam_score"].mean()
plt.bar(avg_score.index, avg_score.values)
plt.title("Average Exam Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.show()
print("Insight:The bar chart compares the average exam scores of different gender groups. The group with the taller bar has the higher average performance.")
```

```python
diet=df["diet_quality"].value_counts()
plt.pie(diet.values,labels=diet.index,autopct="%1.1f%%")
plt.title("Distribution of Diet Quality")
plt.show()
print("Insight:The pie chart shows the proportion of students in each diet quality category. The largest slice represents the most common diet quality among students.")
```

```python
df["Attendance_Range"] = pd.cut(
    df["attendance_percentage"],
    bins=[60, 70, 80, 90, 100],
    labels=["60-69", "70-79", "80-89", "90-100"],
    include_lowest=True
)

attendance = df.groupby("Attendance_Range")["exam_score"].mean()

plt.plot(attendance.index.astype(str), attendance.values, marker="o")
plt.title("Average Score Across Attendance Ranges")
plt.xlabel("Attendance Range")
plt.ylabel("Average Exam Score")
plt.grid(True)
plt.show()
print("Insight:The line chart indicates how average scores change with attendance. An upward trend suggests that better attendance is associated with higher exam scores.")
```
